```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../Data/raw/creditcard.csv")

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
df.info()
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print(
    "Total infinite values:",
    np.isinf(df[numeric_columns]).sum().sum()
)
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]

print(duplicate_rows["Class"].value_counts())
```

```python
print("Class distribution among duplicate rows:")
print(duplicate_rows["Class"].value_counts())
```

```python
print("\nPercentage:")
print(
    duplicate_rows["Class"]
    .value_counts(normalize=True)
    .mul(100)
)
```

```python
print("Target column: Class")
print("\nUnique values:")
print(df["Class"].unique())

print("\nClass distribution:")
print(df["Class"].value_counts())
```

```python
print("\nClass percentage:")
print(
    df["Class"]
    .value_counts(normalize=True)
    .mul(100)
)
```

```python
print("Shape before removing duplicates:", df.shape)

df = df.drop_duplicates().reset_index(drop=True)

print("Shape after removing duplicates:", df.shape)
print("Duplicates remaining:", df.duplicated().sum())
```

```python
X = df.drop("Class", axis=1)
y = df["Class"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
print(y.value_counts())
print("\nPercentage:")
print(y.value_counts(normalize=True) * 100)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
```

```python
# Save original training and test data

X_train.to_csv("../Data/processed/X_train.csv", index=False)
y_train.to_csv("../Data/processed/y_train.csv", index=False)

# X_test and y_test are already saved, so no need to save them again

print("Original training data saved successfully.")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Training set:")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting set:")
print(y_test.value_counts())
print(y_test.value_counts(normalize=True) * 100)
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[["Time", "Amount"]] = scaler.fit_transform(
    X_train[["Time", "Amount"]]
)

X_test_scaled[["Time", "Amount"]] = scaler.transform(
    X_test[["Time", "Amount"]]
)
```

```python
print("Training set:")
print(X_train_scaled[["Time", "Amount"]].describe().loc[["mean", "std"]])

print("\nTesting set:")
print(X_test_scaled[["Time", "Amount"]].describe().loc[["mean", "std"]])
```

```python
print("Training class distribution:")
print(y_train.value_counts())

print("\nTraining class percentage:")
print(y_train.value_counts(normalize=True) * 100)
```

```python
class_counts = y_train.value_counts()

imbalance_ratio = class_counts[0] / class_counts[1]

print("Genuine transactions:", class_counts[0])
print("Fraudulent transactions:", class_counts[1])
print("Imbalance ratio:", round(imbalance_ratio, 2))
```

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scaled,
    y_train
)
```

```python
print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts())
```

```python
print("Original training shape:", X_train.shape)
print("SMOTE training shape:", X_train_smote.shape)

print("Original target shape:", y_train.shape)
print("SMOTE target shape:", y_train_smote.shape)
```

```python
print("Missing values:")
print(X_train_smote.isnull().sum().sum())

print("\nInfinite values:")
print(np.isinf(X_train_smote).sum().sum())
```

```python
print("Number of features:", X_train_smote.shape[1])
print("\nFeatures:")
print(X_train_smote.columns.tolist())
```

```python
print(y_train_smote.value_counts())
```

```python
print("Test set shape:", X_test_scaled.shape)

print("\nTest class distribution:")
print(y_test.value_counts())

print("\nTest class percentages:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
import joblib

joblib.dump(scaler, "../models/scaler.pkl")

print("Scaler saved successfully.")
```

```python
X_train_smote.to_csv(
    "../Data/processed/X_train_smote.csv",
    index=False
)

y_train_smote.to_csv(
    "../Data/processed/y_train_smote.csv",
    index=False
)

X_test_scaled.to_csv(
    "../Data/processed/X_test.csv",
    index=False
)

y_test.to_csv(
    "../Data/processed/y_test.csv",
    index=False
)

print("Processed datasets saved successfully.")
```

```python
# # Preprocessing Summary

# ## Data Cleaning
# - The original dataset was loaded without modifying the raw CSV.
# - Exact duplicate rows were identified and removed.
# - Missing and infinite values were verified.

# ## Feature and Target Separation
# - `Class` was separated as the target variable.
# - `Time`, `V1`–`V28`, and `Amount` were used as input features.

# ## Train-Test Split
# - The data was divided into 80% training and 20% testing data.
# - Stratified splitting was used because of severe class imbalance.

# ## Feature Scaling
# - `Time` and `Amount` were standardized.
# - The scaler was fitted only on the training data and then applied to the test data.
# - The fitted scaler was saved for future inference.

# ## Class Imbalance
# - The dataset contains a severe imbalance between genuine and fraudulent transactions.
# - SMOTE was applied only to the training data.
# - The test set retained the original class distribution.

# ## Final Training Data
# - The SMOTE training dataset contains balanced classes.
# - The test dataset remains representative of the original distribution.

# ## Preprocessing Output
# The processed datasets were saved for model development and evaluation.
```
